# Chapter 1: Is the warehouse task an RL problem?

**Question:** Can "deliver an item, never hit a shelf" be framed as reinforcement learning, and what does a *do-nothing-smart* baseline look like?

**Framing (my own words)**

| RL element | Okanogan warehouse robot | Taxi sandbox (stand-in used for chapters 1-2) |
|---|---|---|
| State | robot position, item position, target position, whether it holds the item, which neighbouring cells are blocked | taxi position, passenger location, destination (500 discrete states) |
| Actions | up, down, left, right | N, S, E, W, pickup, drop-off |
| Reward | -1 per step (be fast), + for pickup/delivery, large penalty for bumping a shelf | -1 per step, +20 delivery, -10 illegal pickup/drop |
| Episode end | item delivered or time limit | passenger delivered or 200 steps |

Taxi is the same problem in miniature: navigate a grid with walls, pick something up, drop it at a target. It is small enough to solve with a table, which makes it the right sandbox to start in.

```python
import sys, json; sys.path.insert(0, "..")
import warnings; warnings.filterwarnings("ignore")
import numpy as np, pandas as pd, matplotlib.pyplot as plt, gymnasium as gym
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .3})
from src.qlearning import evaluate_random
env = gym.make("Taxi-v4")
print("observation space:", env.observation_space)
print("action space:", env.action_space)
```

## Baseline: a random driver

```python
base = evaluate_random(env, episodes=300)
print(base)
json.dump(base, open("../results/taxi_random_baseline.json", "w"))
```

## Takeaways
- A random driver completes only **~3%** of trips (mean return about -776, almost every episode runs the full 200 steps).
- The warehouse task maps one-to-one onto Taxi: same kind of states, moves, step cost, pickup/delivery reward. Anything learned here transfers to the framing of the real problem.
- Taxi is a sandbox, not the warehouse: its 500 states are far fewer than the real task's (see chapter 3).
